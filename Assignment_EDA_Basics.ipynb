# Step 1: Import Libraries and Load Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("House Price India.csv")

# Display first 5 rows
display(df.head())

# Basic information
df.info()
```

The dataset contains **14,620 rows and 23 columns**.


# Step 2: Basic Data Checking

```python
# Check dataset shape
print("Number of rows and columns:", df.shape)

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# Statistical summary
print("\nStatistical Summary:")
display(df.describe())
```

### Explanation

The dataset contains **14,620 records and 23 variables**. There are **no missing values** in the dataset, so no missing-value treatment is required before performing the analysis.


# Question 1: What is the average, median, and standard deviation of house prices?

## Answer

```python
mean_price = df["Price"].mean()
median_price = df["Price"].median()
std_price = df["Price"].std()

print("Average House Price:", round(mean_price, 2))
print("Median House Price:", round(median_price, 2))
print("Standard Deviation:", round(std_price, 2))
```

### Result

* **Average House Price:** 538,932.22
* **Median House Price:** 450,000.00
* **Standard Deviation:** 367,532.38

### Interpretation

The average house price is approximately **538,932**, while the median price is **450,000**.

The average is higher than the median, which suggests that some expensive houses are increasing the overall average price.

The standard deviation is approximately **367,532**, showing that house prices vary considerably across the dataset.


# Question 2: Which number of bedrooms is most common?

## Answer

```python
bedroom_counts = df["number of bedrooms"].value_counts()

print("Number of Houses by Bedroom Count:")
display(bedroom_counts)

most_common_bedrooms = df["number of bedrooms"].mode()[0]

print("Most Common Number of Bedrooms:", most_common_bedrooms)
```

### Result

Three-bedroom houses are the most common in the dataset. This indicates that 3-bedroom properties represent a major portion of the housing market covered by this dataset.


# Question 3: Check if the price data is skewed.

## Answer

```python
price_skewness = df["Price"].skew()

print("Price Skewness:", round(price_skewness, 2))
```

### Result

**Price Skewness = 4.27**

### Interpretation

The price data is **positively skewed (right-skewed)** because the skewness value is significantly greater than zero.

This means that most houses are concentrated around the lower and middle price ranges, while a smaller number of very expensive houses create a long right tail.

This also explains why the average price is higher than the median price.

### Conclusion

The house price distribution is strongly right-skewed, mainly because of some high-priced properties.

# Question 5: What is the relationship between living area and price?

## Answer

For this question, the assignment asks us to use **Area of the house (excluding basement)**.

```python
area_column = "Area of the house(excluding basement)"

correlation = df[area_column].corr(df["Price"])

print("Correlation between House Area and Price:",
      round(correlation, 3))
```

### Result

**Correlation = 0.615**

### Interpretation

There is a **moderate positive relationship** between house area and price.

A correlation of approximately **0.615** means that larger houses generally tend to have higher prices.

However, the relationship is not perfect. This means other factors such as location, number of bedrooms, house grade, waterfront availability, condition, and number of views also influence house prices.

### Conclusion

House area is an important factor affecting price, but it is not the only factor determining the final price.


# Question 6: Identify any anomalies where houses have high prices but low areas.

## Answer

To identify possible anomalies, we can compare houses with very high prices against houses with relatively small living areas.

```python
area_column = "Area of the house(excluding basement)"

price_threshold = df["Price"].quantile(0.95)
area_threshold = df[area_column].quantile(0.25)

anomalies = df[
    (df["Price"] >= price_threshold) &
    (df[area_column] <= area_threshold)
]

print("Number of Possible High-Price / Low-Area Anomalies:",
      len(anomalies))

display(
    anomalies[
        ["Price", area_column, "number of bedrooms",
         "Postal Code", "waterfront present",
         "grade of the house"]
    ].sort_values("Price", ascending=False).head(10)
)
```

### Interpretation

This analysis identifies properties that have unusually high prices even though their house area is relatively small.

Such records may not necessarily be errors. A high price for a smaller house can happen because of:

* Premium location
* Waterfront availability
* High-quality construction
* Better house grade
* Desirable neighborhood
* Other location-related factors

### Conclusion

High-price and low-area properties should be investigated rather than automatically treated as incorrect data.

# Question 9: Which location (zipcode/area) has the highest average price?

## Answer

The `Postal Code` column is used as the location indicator.

```python
postal_average_price = (
    df.groupby("Postal Code")["Price"]
      .mean()
      .sort_values(ascending=False)
)

print("Top 5 Postal Codes by Average House Price:")
display(postal_average_price.head())

highest_postal_code = postal_average_price.idxmax()
highest_average_price = postal_average_price.max()

print("Highest Average Price Postal Code:", highest_postal_code)
print("Average Price:", round(highest_average_price, 2))
```

### Result

**Highest Average Price Postal Code: 122071**

**Average House Price: 2,348,310.81**

### Interpretation

Postal Code **122071** has the highest average house price in the dataset, with an average of approximately **2.35 million**.

This suggests that properties in this location are positioned in a premium segment compared with other postal codes.

### Conclusion

Location is an important factor in determining house prices, and some postal codes have significantly higher property values than others.
